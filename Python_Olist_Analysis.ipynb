```python
orders = pd.read_csv(
    os.path.join(folder, "olist_orders_dataset.csv")
)

order_items = pd.read_csv(
    os.path.join(folder, "olist_order_items_dataset.csv")
)

products = pd.read_csv(
    os.path.join(folder, "olist_products_dataset.csv")
)

customers = pd.read_csv(
    os.path.join(folder, "olist_customers_dataset.csv")
)

print("All datasets loaded successfully!")

print("Orders:", orders.shape)
print("Order Items:", order_items.shape)
print("Products:", products.shape)
print("Customers:", customers.shape)
```

```python
sales_data = order_items.merge(
    products,
    on="product_id",
    how="left"
)

print("Order Items + Products merged!")

print("Shape:", sales_data.shape)

sales_data.head()
```

```python
sales_data = pd.merge(
    sales_data,
    orders,
    on="order_id",
    how="left"
)

print("Orders merged successfully")

print("Final Shape:", sales_data.shape)

display(sales_data.head())
```

```python
# Create a single category column
sales_data["product_category_name"] = (
    sales_data["product_category_name_y"]
    .fillna(sales_data["product_category_name_x"])
)

# Revenue by product category
category_sales = (
    sales_data.groupby("product_category_name")["price"]
    .sum()
    .sort_values(ascending=False)
)

print("Revenue by Product Category:")
display(category_sales.head(10))
```

```python
import pandas as pd
import os

folder = r"C:\Users\merut\OneDrive\Documents\Ecommerce_Business_Analytics"

orders_file = os.path.join(
    folder,
    "olist_orders_dataset.csv"
)

orders_data = pd.read_csv(orders_file)

print("Orders dataset loaded successfully!")
print("Shape:", orders_data.shape)

display(orders_data.head())
```

```python
order_status_count = orders_data["order_status"].value_counts()

print("Order Status Summary:")
display(order_status_count)
```

```python
order_status_count.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sales_with_orders = sales_data.merge(
    orders_data[
        [
            "order_id",
            "order_purchase_timestamp",
            "order_status"
        ]
    ],
    on="order_id",
    how="left"
)

print("Datasets merged successfully!")
print("Shape:", sales_with_orders.shape)

display(sales_with_orders.head())
```

```python
sales_with_orders["order_purchase_timestamp"] = pd.to_datetime(
    sales_with_orders["order_purchase_timestamp"]
)

print("Date column converted successfully!")
```

```python
sales_with_orders["month"] = (
    sales_with_orders["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_revenue = sales_with_orders.groupby(
    "month"
)["price"].sum()

display(monthly_revenue)
```

```python
monthly_revenue.plot(
    kind="line",
    figsize=(12, 6),
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
highest_revenue_month = monthly_revenue.idxmax()
highest_revenue_value = monthly_revenue.max()

lowest_revenue_month = monthly_revenue.idxmin()
lowest_revenue_value = monthly_revenue.min()

print("Highest Revenue Month:", highest_revenue_month)
print("Highest Revenue:", round(highest_revenue_value, 2))

print("\nLowest Revenue Month:", lowest_revenue_month)
print("Lowest Revenue:", round(lowest_revenue_value, 2))
```

```python
cancelled_orders = orders_data[
    orders_data["order_status"] == "canceled"
]

print("Total Cancelled Orders:", len(cancelled_orders))
```

```python
print("========== BUSINESS INSIGHTS ==========")

print(
    "1. Highest revenue category:",
    category_sales.idxmax()
)

print(
    "2. Highest category revenue:",
    round(category_sales.max(), 2)
)

print(
    "3. Total revenue:",
    round(sales_data["price"].sum(), 2)
)

print(
    "4. Average product price:",
    round(sales_data["price"].mean(), 2)
)

print(
    "5. Total orders:",
    len(orders_data)
)

print(
    "6. Cancelled orders:",
    len(cancelled_orders)
)
```
